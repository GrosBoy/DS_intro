# Задание 1. Анализ Titanic
## **Цель**: исследовать набор данных и визуализировать базовые зависимости.
1. Загрузите датасет Titanic (через ссылку или seaborn.load_dataset('titanic')).
2. Постройте **3 разных типа графиков** (на выбор: barplot, histplot, boxplot, countplot, scatterplot).
3. Каждый график должен отвечать на конкретный вопрос, например:
   - Сколько мужчин и женщин было на борту?
   - Как распределяется возраст по классам?
   - Есть ли связь между стоимостью билета и выживаемостью?

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
url="https://raw.githubusercontent.com/pandas-dev/pandas/master/doc/data/titanic.csv"
ds=pd.read_csv(url)
print('Loaded dataset')
ds['age_filled']=ds['Age'].fillna(ds['Age'].median())
ds['Sex_U'] = ds['Sex'].map({'male': 0,'female': 1})
count_sex=ds['Sex'].value_counts()
sns.set(style='whitegrid')
sns.countplot(x='Sex',data=ds)
plt.title("Кол-во мужчин и женщин")
plt.show()
```

```python
sns.histplot(data=ds,x='Age',hue='Pclass',kde=True)
plt.title('Распределение возраста по классам')
```

```python
sns.boxplot(data=ds,x='Survived',y='Fare')
plt.show()
```

Ответы на вопросы:


1. Мужчин было больше, чем женщин.
2. Больше всего молодых людей было в третьем классе. В первом классе в основном были люди среднего и пожилого возраста. Во втором классе были люди в возрасте от 25 до 35 лет.
3. Те, кто заплатил меньше, чаще умирали, а те, кто заплатил больше, имели больше шансов выжить.
